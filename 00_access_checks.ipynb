```python
import requests

SHOP = "team-agentic-flow"
cid = dbutils.secrets.get("agenticflow", "shopify_client_id")
secret = dbutils.secrets.get("agenticflow", "shopify_client_secret")

tok = requests.post(
    f"https://{SHOP}.myshopify.com/admin/oauth/access_token",
    data={"grant_type": "client_credentials", "client_id": cid, "client_secret": secret},
    timeout=30,
)
print("Token step:", tok.status_code)
tok.raise_for_status()
token = tok.json()["access_token"]

r = requests.post(
    f"https://{SHOP}.myshopify.com/admin/api/2025-10/graphql.json",
    json={"query": "{ products(first: 5) { edges { node { title totalInventory } } } }"},
    headers={"X-Shopify-Access-Token": token},
    timeout=30,
)
print(r.json())
```
